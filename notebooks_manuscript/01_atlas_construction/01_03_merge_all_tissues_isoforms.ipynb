```python
import logging
logging.getLogger("fontTools").setLevel(logging.WARNING)

import os
_r = os.path.abspath(".")
while _r != os.path.dirname(_r) and not os.path.exists(os.path.join(_r, ".notebooks_root")):
    _r = os.path.dirname(_r)
os.chdir(_r if os.path.exists(os.path.join(_r, ".notebooks_root")) else "/oak/stanford/groups/quake/mmantri/group.quake/tabula_longread/notebooks")  # cd to notebooks/ root (marked .notebooks_root); relocation-proof

import scanpy as sc
import seaborn as sns
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import anndata as ad
from collections import Counter
from scipy.sparse import csr_matrix 
import gc
import os, sys
import glob
gc.enable()

sys.path.insert(0, os.path.dirname(os.path.abspath("__file__")))
from add_classification import add_classification
```

```python
sc.settings.verbosity = 3  
# verbosity: errors (0), warnings (1), info (2), hints (3)
sc.settings.set_figure_params(dpi=80, facecolor='white', frameon=True, figsize=(5, 5))
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42
```

```python
BASEDIR = "./../pacbio/recollapsed/"

# When True, add the SQANTI3 RulesFilter classification at the MOLECULE (PB.X.Y)
# level for each tissue and drop molecules whose filter_result == "Artifact"
# BEFORE aggregating to ensemblids/ensemblids_annotatedonly. This removes SQANTI artifact
# molecules from the pbids, ensemblids AND ensemblids_annotatedonly matrices (and from the
# QC / min_counts cell-calling, since it runs before those steps).
sqanti_filter = False

sample_metadata = pd.read_csv("./../csvs/sample_metadata.csv")
donors = sorted(sample_metadata["donor"].dropna().unique())
donors = [d for d in donors if d != "TSP31"]  # exclude donor TSP31

os.makedirs("./../pacbio/h5ads", exist_ok=True)

for donor in donors:
    print(f"\n=== {donor} ===")
    donor_tubes = sorted(
        sample_metadata.loc[sample_metadata["donor"] == donor, "tube_id"].dropna().unique()
    )
    adata_dict = dict()
    adata_dict_ensemblids = dict()

    for tissue in donor_tubes:
        seurat_dir = f"{BASEDIR}/seurat_monoexon/{tissue}/isoforms_seurat/"
        if not os.path.isdir(seurat_dir):
            print(f"Skipping {tissue}: {seurat_dir} not found")
            continue
        print(tissue)
        adata = sc.read_10x_mtx(seurat_dir)

        # Build transcript map from pigeon annotated.info.csv
        annot = pd.read_csv(
            f"{BASEDIR}/seurat_monoexon/{tissue}/{tissue}.annotated.info.csv",
            sep="\t", usecols=["pbid", "transcript"]
        ).drop_duplicates(subset="pbid")
        # Use ENST ID if available, otherwise keep the PB ID (novel)
        annot["transcript_id"] = annot["transcript"].where(
            annot["transcript"].astype(str).str.startswith("ENST"),
            annot["pbid"]
        )
        transcript_map = annot.set_index("pbid")

        adata.var[["molecule_id", "gene_name"]] = adata.var["gene_ids"].str.split(":", expand=True)
        adata.var_names = adata.var.molecule_id
        adata.var["transcript_id"] = transcript_map.loc[adata.var.molecule_id, "transcript_id"].values

        # # Add pigeon classification metadata (prefix="pigeon_" to avoid column name conflicts)
        # add_classification(
        #     adata,
        #     f"{BASEDIR}/classify/{tissue}/collapse_classification.filtered_lite_classification.txt",
        #     prefix="pigeon_",
        # )

        # Optionally add SQANTI3 classification at the molecule level and drop
        # SQANTI "Artifact" molecules, so they are excluded from the molecules,
        # ensemblids and ensemblids_annotatedonly matrices (aggregation below sums only the
        # surviving molecules per transcript_id).
        if sqanti_filter:
            sqanti_path = f"{BASEDIR}/sqanti3/{tissue}/{tissue}_RulesFilter_classification.txt"
            if os.path.exists(sqanti_path):
                add_classification(adata, sqanti_path, prefix="sqanti_")
                if "sqanti_filter_result" in adata.var.columns:
                    keep_mask = adata.var["sqanti_filter_result"].values != "Artifact"
                    n_before = adata.n_vars
                    adata = adata[:, keep_mask].copy()
                    print(f"  SQANTI filter: dropped {n_before - adata.n_vars} artifact molecules ({adata.n_vars} kept)")
                else:
                    print(f"  SQANTI filter: 'filter_result' column absent in {sqanti_path}; no molecules dropped")
            else:
                print(f"  SQANTI filter: {sqanti_path} not found; skipping SQANTI filter for {tissue}")

        # Cap percent_top to number of features (tissues like T4 can have < 500)
        n_vars = adata.n_vars
        percent_top = [n for n in [50, 100, 200, 500] if n <= n_vars]
        sc.pp.calculate_qc_metrics(adata, percent_top=percent_top, inplace=True, log1p=True)
        print(adata.shape)
        data = adata.obs[["total_counts"]].sort_values(by="total_counts", ascending=False)
        data["cell_number"] = [i for i in range(data.shape[0])]
        sc.settings.set_figure_params(dpi=150, dpi_save=300, fontsize=8, frameon=True, figsize=(3, 3), transparent=False, format="pdf")
        fig, ax = plt.subplots(figsize=(4.0, 3.0))
        plt.rc('font', size=6)
        ax.set_axisbelow(True)
        plt.grid(False)
        plt.xlabel("Droplets")
        plt.ylabel("UMI counts")
        sns.scatterplot(data=data, x="cell_number", y="total_counts", linewidth=0, size=1.0, rasterized=True, legend=None, ax=ax)
        plt.xscale('symlog')
        plt.yscale('symlog')
        plt.axhline(500)
        plt.show()
        adata.obs["tube_id"] = tissue
        print("Pre-filtering: ", adata.shape)
        sc.pp.filter_cells(adata, min_counts=500)
        print("Post-filtering: ", adata.shape)
        adata_dict[tissue] = adata
        adata_dict_ensemblids[tissue] = sc.get.aggregate(adata, by="transcript_id", func="sum", axis=1)
        adata_dict_ensemblids[tissue].X = csr_matrix(adata_dict_ensemblids[tissue].layers["sum"])
        del adata_dict_ensemblids[tissue].layers["sum"]
        mapping_dict = dict(zip(adata_dict[tissue].var['transcript_id'], adata_dict[tissue].var['gene_name']))
        adata_dict_ensemblids[tissue].var['gene_name'] = adata_dict_ensemblids[tissue].var['transcript_id'].map(mapping_dict)

    if not adata_dict:
        print(f"  No tissues found for {donor}, skipping write")
        continue

    # ---- pbids (PB.X.Y level) ----
    all_var = pd.concat([a.var for a in adata_dict.values()], join="outer")
    all_var_dedup = all_var[~all_var.index.duplicated()]
    adata = ad.concat(adatas=adata_dict.values(), join="outer", keys=adata_dict.keys(), index_unique="_")
    adata.var = all_var_dedup.loc[adata.var_names]
    adata.var.index = adata.var.index.values
    adata.obs = adata.obs.join(sample_metadata.set_index("tube_id"), on="tube_id")
    for col in adata.var.columns:
        if adata.var[col].dtype == object:
            adata.var[col] = adata.var[col].fillna("")
    out_mol = f"./../pacbio/h5ads/{donor}_pacbio_recollapsed_raw_counts_pbids_bc_anndata_mincounts500_monoexon.h5ad"
    adata.write_h5ad(out_mol)
    print(f"  pbids: {adata.shape} -> {out_mol}")
    del adata, all_var, all_var_dedup
    gc.collect()

    # ---- ensemblids (ENST when available, else PB.X.Y) ----
    all_var = pd.concat([a.var for a in adata_dict_ensemblids.values()], join="outer")
    all_var = all_var[~all_var.index.duplicated()]
    adata_ensemblids = ad.concat(adatas=adata_dict_ensemblids.values(), join="outer", keys=adata_dict_ensemblids.keys(), index_unique="_")
    adata_ensemblids.var = all_var.loc[adata_ensemblids.var_names]
    adata_ensemblids.var.index = adata_ensemblids.var.index.values
    adata_ensemblids.obs = adata_ensemblids.obs.join(sample_metadata.set_index("tube_id"), on="tube_id")
    out_tr = f"./../pacbio/h5ads/{donor}_pacbio_recollapsed_raw_counts_ensemblids_bc_anndata_mincounts500_monoexon.h5ad"
    adata_ensemblids.write_h5ad(out_tr)
    print(f"  ensemblids: {adata_ensemblids.shape} -> {out_tr}")

    # ---- ensemblids_annotatedonly (ENST-only subset of ensemblids) ----
    adata_ensemblids_annotatedonly = adata_ensemblids[:, adata_ensemblids.var["transcript_id"].str.startswith("ENST")]
    out_iso = f"./../pacbio/h5ads/{donor}_pacbio_recollapsed_raw_counts_ensemblids_annotatedonly_bc_anndata_mincounts500_monoexon.h5ad"
    adata_ensemblids_annotatedonly.write_h5ad(out_iso)
    print(f"  ensemblids_annotatedonly: {adata_ensemblids_annotatedonly.shape} -> {out_iso}")

    del adata_dict, adata_dict_ensemblids, adata_ensemblids, adata_ensemblids_annotatedonly, all_var
    gc.collect()
```
